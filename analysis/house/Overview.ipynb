# Alaska 2026 — House

Saved November 11 FairVote results. This overview reads aggregate files only; it does not rerun MDS. The historical scikit-learn 1.6.1 solver has an unresolved version issue; see [known issues](../../docs/KNOWN_ISSUES.md).

See this folder’s README for rankings, transfers and other race-specific work.

```python
from pathlib import Path
import json, sys
root = next(p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (p / 'analysis/shared/paths.py').is_file())
sys.path.insert(0, str(root))
from analysis.shared.paths import model_dir
from IPython.display import display, HTML
import html
folder = model_dir('us_house')
result = json.loads((folder / 'mds_results.json').read_text())
print('Source:', result['source_filename'])
print('Source SHA-256:', result['source_sha256'])
print('Source records:', f"{result['source_rows']:,}")
print('Saved solver:', result['environment']['sklearn'])
```

```python
import csv
axis = list(csv.DictReader((folder / 'candidate_axis.csv').open()))
column = 'display_axis_position' if 'display_axis_position' in axis[0] else 'axis_position'
rows = sorted(axis, key=lambda row: float(row[column]))
display(HTML('<table><tr><th>Candidate</th><th>Axis coordinate</th></tr>' + ''.join('<tr><td>' + html.escape(row['candidate']) + '</td><td>' + f"{float(row[column]):.3f}" + '</td></tr>' for row in rows) + '</table>'))
display(HTML('<table><tr><th>Scored group</th><th>N</th><th>Gamma</th><th>Random mean</th></tr>' + ''.join('<tr><td>' + html.escape(row['group']) + '</td><td>' + f"{row['ballots']:,}" + '</td><td>' + f"{100*row['observed_consistency']:.2f}%" + '</td><td>' + f"{100*row['random_mean']:.2f}%" + '</td></tr>' for row in result['summary'][:2]) + '</table>'))
print('These checks describe saved outputs, not a corrected-solver rerun or certified election totals.')
```
